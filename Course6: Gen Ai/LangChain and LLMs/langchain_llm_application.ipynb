## Model I/O: 
- prompts: serve as inputs for a language model and guide it on what to generate
- Language Models: take the prompts and generate text based on their training
- Output parsers: transform the raw output form the language model into a format that is useful and accessible.

## Text Splitters
- splitting text and documents in Langchain, for storage and segmentation
    - CharacterTextSplitter:
        - in langChain divides large text into smaller sections using a defined set of separator characters to determine the chunck size
        - The chunck size of the characterTextSplitter determines the maximum number of characters in each chunk of text

    - RecursiveTextSplitter: 
        - in langChain recursively splits text into chunks for better handling, unlike characterTextSplitter, which has fixed splitting approach
        - There are different types of text splitters avaiable in LangChain, and you can specficy parameters such as chunk size and chunk overlap

## Embeddings in GenAi
- In genAi, embeddings conver complex data into compact vector forms. These vectors, usually floating point numbers, encapsulate the characteristics of the dataset. Vector distances indicate relatedness: smaller distances suggest a higher correctlation. 
- Embeddings represent text, images, and more in a machine friendly manner. They help Ai grasp data meaning naturally, leading to text generation, language translation, and image understanding applications. 

## Text embedding Models in LangChain
- LangChain provides a standardized interface for various embedding model providers like OpenAi, Cohere, and hugging Face

- The models transform text into vector representations, enabling operations like semantic search through text similarity in vector space. 

- To begin text embedding models, install the required packages and set up API keys. For OpenAi in langchain, these steps are already complete

- the `embed_docuements` method embeds multiple texts, providing a list of vector representations. 

## VectorStore
- VectorStore acts as storage space for embeddings. It helps quickly find and work with data representations, improving the model's performance

- The vectorStore in LangChain is an abstract class that represents a store of vectors
- It is utilized for adding vectors and documents to the store, removing items from the store, and conducting searches within the store
- It takes care of storing embedded data and performing vector searches for you
- There are different implementations of VectorStore based on different technologies such as ChromaDB, Faiss, pgVector databases, and pinecone

## How VectorStore Works
- process of storing and searching over unstructured data works in the context of LangChain's VectorStore

    - Embedding data:
        - Embeddings are used to transform unstructured formats.Storing these resulant vectors is used for handling such data efficiently

    - Storing Vectors:
        - the VectorStore in LangChain takes care of storing the embedded data

    - Embedding query:
        - At the time of querying, the unstructured query is transformed into an embedded form

    - Retrieving Vector:
        - The VectorStore retrieves the embedding vectors that are similar to the embedding query

    - Vector Search:
        - VectorStore conducts a vector search, finding the most similar vectors to a given one. 

## LangChain Retriever
- LangChain retrievers are tools in LangChain that fetch information from external sources to enhance the language model's knowledge and capabilities. 
- LangChain retrievers fetch documents with unstructured queries, offering versatility beyond vectore stores, which primarily store and retrieve.
- While vector stores can form the backbone of a retriever, there are other types of retrievers as well.
- Retrievers take a string query as input and return a list of document objects as output. 

## LangChain Chains
- using a LLM alone is suitable for basic taks, but complex applications may require chaining LLMs or other components
- LangChain offers two high-level frameworks for the chaining of components:

    - Chain interface: 
        - this the legacy approach for chaining components. 

    - LangChain expression language (LCEL):
        - This is the updated approach recommended for chain composition in new applications

## Foundational LangChain Chain
- In applications of LLMs, chaining typically involves integrating a prompt template with the LLM and, if necessary, employing an output parser for futher refinement

- An LLMChain, commonly used in LangChain, boosts language model functionality and is integrated into other chains and agents.

- An LLMChain consists of a promptTemplate and a language model 

- The LLMChain formats the prompt template using input and memeory key values, then passes it to the LLM for output
