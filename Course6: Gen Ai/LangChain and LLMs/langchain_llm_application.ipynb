## LangChain for LLM application Development

## Model I/O: prompts, language models, and parsers
# Flow of Chatbot Application
- Large Document
- Broken into smaller chunks
- Chuncks stored into vector store
- Retrieve relevant info
- create completion with relevant info
- get reply

## Model I/O: 
- prompts: serve as inputs for a language model and guide it on what to generate
- Language Models: take the prompts and generate text based on their training
- Output parsers: transform the raw output form the language model into a format that is useful and accessible. 

## Document Loaders
- first step is to import the raw document and explore the process of importing documents with examples

- CSV document loader: file is a type of text file that uses comma as a delimiter to seperate values.

- file directory loader: is like a filing cabinet where organized files are stored, managing data. 

- HTML document loader: is a standard markup language used for creating documents intended for web brower display

- JSON document loader: is a standard file format and data interchange format that employs human readable text to store and transmit data objects

- PDF document loader: is a file format disgned to display documents, including text formatting and images

## Text Splitters
- splitting text and documents in Langchain, for storage and segmentation
    - CharacterTextSplitter:
        - in langChain divides large text into smaller sections using a defined set of separator characters to determine the chunck size
        - The chunck size of the characterTextSplitter determines the maximum number of characters in each chunk of text

    - RecursiveTextSplitter: 
        - in langChain recursively splits text into chunks for better handling, unlike characterTextSplitter, which has fixed splitting approach
        - There are different types of text splitters avaiable in LangChain, and you can specficy parameters such as chunk size and chunk overlap

## Embeddings in GenAi
- In genAi, embeddings conver complex data into compact vector forms. These vectors, usually floating point numbers, encapsulate the characteristics of the dataset. Vector distances indicate relatedness: smaller distances suggest a higher correctlation. 
- Embeddings represent text, images, and more in a machine friendly manner. They help Ai grasp data meaning naturally, leading to text generation, language translation, and image understanding applications. 

## Text embedding Models in LangChain
- LangChain provides a standardized interface for various embedding model providers like OpenAi, Cohere, and hugging Face

- The models transform text into vector representations, enabling operations like semantic search through text similarity in vector space. 

- To begin text embedding models, install the required packages and set up API keys. For OpenAi in langchain, these steps are already complete

- the `embed_docuements` method embeds multiple texts, providing a list of vector representations.
