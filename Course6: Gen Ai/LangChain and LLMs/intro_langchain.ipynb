## Introduction to LLMs and LangChain

## Language Models
- A language model is a probabilistic machine learning entity.
- it resembles a complex function, designed to predict the probability of words sequences within a specific language corpus. 
- The Language Models operate by assigning probabilities to sequences of words
    - Consider the sentence: "This is a new technology"
    - The language model calculates the probability of the sentence as:
        - P(This is a new technology)
        - P(This is a new technology) = P(This) P(is|This) P(a|This is) P(new|This is a) P(technology| This is a new)
- rare and a challenging words result in a lower probability

## Large Language Models
- Large Language Models (LLMs) are state of the art AI models designed to comprehend and generate human language.
- Large: Refers to the significant size and complexity of these models, which contains hundreds of millions or even billions of parameters
- Language: Denotes their primary function, which is to understand and generate human language
- Model: Describes them as mathematical representations that capture the patterns and structre of language data. 

## components of LLMs

`Tokenization`: This process involves breaking down text into smaller units called tokens, which can be words, phrases, or even individual characters. 

`Embedding`: This embedding component maps tokens to a high dimensional vector space, representing each token with a unique vector. 

`Attention`: The attention mechanism lets the model concentrate on specific parts of the input text when generating output. 

`Pretraining`: Pretraining in LLMs involves training them on large volumes of text data to capture and learn the intricate patterns and structures inherent in human language. 

`Transfer Learning`: This component allows the model to adapt to new tasks by fine tuning the pretrained model on a smaller dataset

`Encoder and Decoder`: This employs the Transformer framework in a large language model architecture, comprising two main parts: an encoder and a decoder. 

`Scaling`: This necessitates significant computational resources for training and upkeep, making scaling a challenging but essential part of its architecture

## Functions of components within LLM architecture

`Input embeddings`: The input text is convered into a format the model understands, called embeddings. These embeddings are high dimensional vectors that capture the semantic properties of words. 

`Positional encoding`: since the model does not inherently understand the order of words in a sentence, positional encodings are added. It is added to the input embeddings to provide contextual information about the position of words within the sentence

`Encoder`: The encoder processes the input embeddings through multiple layers.
- Attention mechanism: Each layer applies self attention mechanisms that allow the model to weigh the importance of different words relative to each other. 
- Feed forward: it is followed by feed forward neural networks for further transformation. 

`Decoder`: In models that include a decoder, the decoder generates output text by processing the encoder's output along with its previous outputs. The decoder helps the machine predict what word comes next based on what it has understood so far. 

`Multi head attention`: This mechanism allows the model to focus on different positions of the input sequence simultaneously. It splits the attention mechanism into multiple 'heads', each capable of capturing different types of dependencies in the input data. 

`Layer normalization`: This is a technique to standardize the inputs to each layer within the network, it stabilizes the learning process and improves the speed of training. 

`Output`: Finally, the model produces its own understanding or generates new sentences. The output depends on what the model is designed to do. For example, if its predicting the next word in a sentence, it gives a probability for each word.
