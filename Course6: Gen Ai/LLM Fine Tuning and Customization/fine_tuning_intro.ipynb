## LLM Fine Tuning and Customization

- LLM fine tuning invovles refining a language model for specific tasks using a curated dateset
    - Fine tuning LLM means customizing them for specific tasks using labeled data that fits the job. 
    - Fine tuning helps LLMs understand the connections between inputs and outputs in a specific dataset, refining their predictions
    - The fine tuning process can be time consuming and requires a good dataset for fine tuning, but it makes LLMs more efficient and effective for specialized tasks. 

## Need of Fine Tuning
- Fine tuning LLMs is crucial to make them fit specific taks, improving how well they work and relate to the context
- Its useful to make LLMs specific to a domain, handle rare cases, and keep data private and secure
- Even with challenges, fine tuning is vital for tailoring LLMs to specific needs, esecially for organizations using them in applications. 

## Data preparation for LLM tuning
- Data preparation is vital for fine tuning LLMs to ensure the model excels in specific tasks. 
- The LLM fine tuning process has two phases:
    - Data preparation
    -   Fine Tuning

## Fine Tuning Methodologies
- Fine tuning methodologies for LLMs can be categorized into two approaches:
- supervised fine tuning:
    - It customizes the model for a task with labeled data, the target model borrows designs from the source model, tweaking parameters except the output layer. this boosts transfer learning and accuracy. 
- Reinforcement learning with human feedback:
    - The RLHF method aims for gradual improvements, trying strategies like regularization or changing the model structure. Engineers can refine the model iteratively until it reaches the desired performance level. 

## Supervised fine tuning
- supervised fine tuning is like giving a model extra traning with specific examples and clear labels. It helps the model get a better at understanding and predicting things accurately. 

## parameter efficient fine tuning
- parameter efficient fine tuning is a technique used in NLP to improve the performance of pretraining language models on specific tasks while significantly reducing computational requirements. 
- it enables fine tuning a small subset of parameters in a pretrained LLM, making it more efficient and cost effective

- LoRA (low rank adapters): 
    - It introduces traninable parameters into the transformer layers to enable the model to learn specific representations. 

- P tuning (prompt tuning): 
    - it adds a learnable prompt to the input to help the model adapt to new tasks. 

- prefix tuning:
    - It enhances taks specific learning in transformers by adding trainable tensorsto each block 

- Adapters:
    - They add tunable layers to LLMs transformer blocks for task learning

- AdaLoRA:
    - it extends LoRA, enabling fine tuning on quantized weights for efficient training.

## Reinforcement Learning with Human Feedback
- RLHF is about training LLMs using human feedback, helping them follow instructions, and providing quality judgements. type of machine learning where a model is trained and adjusted based on feedback provided by humans.

## Hyperparameter tuning in fine tuning
- Hyperparameter adjusting is essential when fine tuning language models for optimal performance. These parameters are set before the model learns, helping to shape its performance. 

- Fine tuning includes adjusting key hyperparameters like learning rate, batch size, regularization strength, and number of epochs. 
- Tuning hyperparameters means finding the best combo for effective learning without sacrificing speed or risking overfitting
- Hyperparameter tuning methods vary, from basic manual adjustments to advanced approaches like Bayesian optimization and grid search.
- The method chosen depends on the task, data, and resources. It is crucial for LLM fine tuning, impacting performance. 

## Evaluation of Fine Tuning Model
- The evaluation of a fine tuned model is a critical step in determining its performance and effectiveness for a specific taks or domain. 
- involves the meticulous process of refining a pretrained model to better align with a specific task
- It is the process of adjusting the parameters of a pretrained model to a spcific taks or domain

## Hands on Fine Tuning
- the leading LLM fine tuning frameworks and libraries that elevate language model's performance and adaptability for specific taks and industries are:
    - HuggingFace Accelerate:
        - a popular library for model parallel training and inference, which can be used in conjunction with DeepSpeed for fine tuning LLMs

    - HuggingFace Transformers:
        - offers comprehensive framework for fine tuning LLMs, including support for LoRA and other parameter efficient fine tuning techniques. Provides pretrained models, datasets, and tools for fine tuning, making it easy for developers to adapt models to specific tasks.

    - DeepSpeed:
        - high performance computing library that can be used for distributed traning and model parallel training, which is particularly useful for fine tuning large models like code Llama, integrating well with huggingFaces's transformers library. 

    - SuperAnnotate:
        - It is a platform that offers tools and services for fine tuning LLMs, including data preparation, hyperparameter tuning, and model evaluation. 

## Fine tuning best practices
- Data preparation:
    - Label and format data consistently, employing augmentation techniques for a diverse training dataset

- Model selection:
    - Pick tools like huggingFace or DeepSpeed to optimize fine tuning

- Monitor overfitting:
    - Watch for overfitting in small datasets, use techniques liek early stopping

- Hyperparameter tuning:
    - Use Bayesian or hyperband optimization to find optimal hyperparameters efficiently

- Bias awareness:
    - Watch for biases in pretrained models and use techniques to mitigate them

- Model evaluation:
    - evaluate the model using metrics to gauge effectiveness for the intended application

## Common Biases with LLM fine tuning
- pretrained models may have built in biases in what they've learned, which continue during the fine tuning process.

    - spurious correlations and underrepreentation:
        - Biases in data, like false connections or uneven representation, can carry over during fine tuning.

    - fairness and inductive biases:
        - To predict a feature's impact, consider how easily it can be extracted after pretraining and the evidence during fine tuning.
